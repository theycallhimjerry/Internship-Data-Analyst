```python
#Import libraries
import pandas as pd
import numpy as np
from pathlib import Path

pd.set_option("display.max_columns", 100)
pd.set_option("display.max_rows", 100)
print("Libraries imported successfully.")
print("pandas:", pd.__version__, "| NumPy:", np.__version__)
```

```python
#Locate and load the datasets
file_paths = {
    "customer": "/Users/jerry/Desktop/INTERNSHIP_SUITE/CSV/Customer_Master_Snapshot.csv",
    "sales": "/Users/jerry/Desktop/INTERNSHIP_SUITE/CSV/Sales_Transaction_Cleaned.csv",
    "events": "/Users/jerry/Desktop/INTERNSHIP_SUITE/CSV/Customer_Behaviour_Events.csv"
}

def locate_file(filename):
    for path in [Path(filename), Path("/mnt/data") / filename]:
        if path.exists():
            return path
    raise FileNotFoundError(
        f"Cannot find {filename}. Put it beside this notebook or edit file_paths."
    )

resolved_paths = {name: locate_file(filename)
                  for name, filename in file_paths.items()}

customer = pd.read_csv(resolved_paths["customer"])
sales = pd.read_csv(resolved_paths["sales"])
events = pd.read_csv(resolved_paths["events"])

for name, path in resolved_paths.items():
    print(f"{name}: {path}")
print("\nDimensions:")
print("Customer master:", customer.shape)
print("Sales transactions:", sales.shape)
print("Behaviour events:", events.shape)
```

```python
#Inspect columns, data types and sample records
for name, df in {
    "Customer master": customer,
    "Sales transactions": sales,
    "Behaviour events": events
}.items():
    print(f"\\n--- {name} ---")
    print("Columns:", df.columns.tolist())
    display(df.head())
    df.info()
```

```python
#Missing value assesment
def missing_value_report(df):
    report = pd.DataFrame({
        "Missing Values": df.isna().sum(),
        "Missing Percentage": (df.isna().mean() * 100).round(2)
    })
    return report.sort_values("Missing Percentage", ascending=False)

for name, df in {
    "Customer master": customer,
    "Sales transactions": sales,
    "Behaviour events": events
}.items():
    print(f"\\n{name}")  
    display(missing_value_report(df))
```

```python
#Duplicate records and candidate key uniqueness
def key_check(df, key, dataset_name):
    if key not in df.columns:
        return {"Dataset": dataset_name, "Candidate Key": key,
                "Column Exists": False, "Rows": len(df),
                "Unique Values": np.nan, "Duplicate Key Values": np.nan,
                "Null Key Values": np.nan}
    return {"Dataset": dataset_name, "Candidate Key": key,
            "Column Exists": True, "Rows": len(df),
            "Unique Values": df[key].nunique(dropna=True),
            "Duplicate Key Values": int(df[key].duplicated().sum()),
            "Null Key Values": int(df[key].isna().sum())}

duplicate_summary = pd.DataFrame([
    {"Dataset": name, "Duplicate Entire Rows": int(df.duplicated().sum())}
    for name, df in {
        "Customer master": customer,
        "Sales transactions": sales,
        "Behaviour events": events
    }.items()
])

candidate_keys = pd.DataFrame([
    key_check(customer, "customer_id", "Customer master"),
    key_check(sales, "transaction_id", "Sales transactions"),
    key_check(events, "event_id", "Behaviour events")
])
display(duplicate_summary)
display(candidate_keys)
```

```python
#Compare customer identifiers and assess potential relationships
for name, df in {
    "Customer master": customer,
    "Sales transactions": sales,
    "Behaviour events": events
}.items():
    if "customer_id" not in df.columns:
        raise KeyError(f"'customer_id' is missing from {name}")

customer_ids = set(customer["customer_id"].dropna().astype(str))
sales_ids = set(sales["customer_id"].dropna().astype(str))
event_ids = set(events["customer_id"].dropna().astype(str))

relationship_summary = pd.DataFrame([
    {"Comparison": "Customer master vs sales",
     "Matching distinct customer IDs": len(customer_ids & sales_ids)},
    {"Comparison": "Customer master vs behaviour events",
     "Matching distinct customer IDs": len(customer_ids & event_ids)},
    {"Comparison": "Sales vs behaviour events",
     "Matching distinct customer IDs": len(sales_ids & event_ids)}
])
display(relationship_summary)

sales_not_in_events = sales_ids - event_ids
events_not_in_sales = event_ids - sales_ids
sales_rows_matching_events = sales[
    sales["customer_id"].astype(str).isin(event_ids)
]

print("Distinct sales IDs absent from events:", len(sales_not_in_events))
print("Distinct event IDs absent from sales:", len(events_not_in_sales))
print("Sales rows whose customer ID occurs in events:",
      len(sales_rows_matching_events))
print("Percentage of sales rows matching an event customer:",
      round(100 * len(sales_rows_matching_events) / len(sales), 2)
      if len(sales) else np.nan)

unmatched_ids = pd.concat([
    pd.DataFrame({"customer_id": sorted(sales_not_in_events),
                  "source_dataset": "Sales transactions",
                  "issue": "Customer ID not found in behaviour events"}),
    pd.DataFrame({"customer_id": sorted(events_not_in_sales),
                  "source_dataset": "Behaviour events",
                  "issue": "Customer ID not found in sales transactions"})
], ignore_index=True)
display(unmatched_ids.head(20))
```

```python
#Date profiling and chronological checks
date_columns = {
    "customer": ["snapshot_date", "registration_date", "last_activity_date"],
    "sales": ["transaction_date"],
    "events": ["event_date"]
}

def convert_date_column(df, column):
    if column in df.columns:
        df[column] = pd.to_datetime(df[column], errors="coerce")
    else:
        print(f"Column not found; skipping: {column}")

for col in date_columns["customer"]:
    convert_date_column(customer, col)
for col in date_columns["sales"]:
    convert_date_column(sales, col)
for col in date_columns["events"]:
    convert_date_column(events, col)

def date_range_report(df, dataset_name, columns):
    rows = []
    for col in columns:
        if col in df.columns:
            rows.append({
                "Dataset": dataset_name, "Date Column": col,
                "Valid Dates": int(df[col].notna().sum()),
                "Missing_or_invalid": int(df[col].isna().sum()),
                "Earliest Date": df[col].min(),
                "Latest Date": df[col].max()
            })
    return rows

date_ranges = pd.DataFrame(
    date_range_report(customer, "Customer master", date_columns["customer"]) +
    date_range_report(sales, "Sales transactions", date_columns["sales"]) +
    date_range_report(events, "Behaviour events", date_columns["events"])
)
display(date_ranges)

date_issue_frames = []
if {"registration_date", "snapshot_date"}.issubset(customer.columns):
    issue = customer[
        customer["registration_date"].notna() &
        customer["snapshot_date"].notna() &
        (customer["registration_date"] > customer["snapshot_date"])
    ].copy()
    issue["issue"] = "Registration date after snapshot date"
    date_issue_frames.append(issue)

if {"last_activity_date", "registration_date"}.issubset(customer.columns):
    issue = customer[
        customer["last_activity_date"].notna() &
        customer["registration_date"].notna() &
        (customer["last_activity_date"] < customer["registration_date"])
    ].copy()
    issue["issue"] = "Last activity date before registration date"
    date_issue_frames.append(issue)

date_issues = (pd.concat(date_issue_frames, ignore_index=True)
               if date_issue_frames else pd.DataFrame())
print("Chronological issue records:", len(date_issues))
display(date_issues.head(20))
```

```python
#Check the recorded tenure calculation
tenure_mismatch = pd.DataFrame()
required = {"snapshot_date", "registration_date", "tenure_days"}

if required.issubset(customer.columns):
    tenure_check = customer.copy()
    tenure_check["calculated_tenure_days"] = (
        tenure_check["snapshot_date"] - tenure_check["registration_date"]
    ).dt.days
    tenure_check["tenure_difference"] = (
        pd.to_numeric(tenure_check["tenure_days"], errors="coerce") -
        tenure_check["calculated_tenure_days"]
    )
    tenure_mismatch = tenure_check[
        tenure_check["tenure_difference"].notna() &
        (tenure_check["tenure_difference"] != 0)
    ].copy()
    print("Rows tested:", len(tenure_check))
    print("Rows matching assumed calculation:",
          int((tenure_check["tenure_difference"] == 0).sum()))
    print("Rows not matching assumed calculation:", len(tenure_mismatch))
    display(tenure_mismatch.head(20))
else:
    print("Required tenure/date columns not all present; check column names.")
```

```python
#Numerical and categorical profiling
for name, df in {
    "Customer master": customer,
    "Sales transactions": sales,
    "Behaviour events": events
}.items():
    print(f"\\n--- Numerical profile: {name} ---")
    display(df.describe(include=[np.number]).T)

def categorical_profile(df, dataset_name, max_categories=30):
    rows = []
    for col in df.select_dtypes(include=["object", "category"]).columns:
        counts = df[col].value_counts(dropna=False)
        for value, count in counts.head(max_categories).items():
            rows.append({
                "Dataset": dataset_name, "Column": col,
                "Category": str(value), "Count": int(count)
            })
    return rows

categorical_summary = pd.DataFrame(
    categorical_profile(customer, "Customer master") +
    categorical_profile(sales, "Sales transactions") +
    categorical_profile(events, "Behaviour events")
)
display(categorical_summary.head(50))
```

```python
#Investigate the sales revenue calculation
revenue_check = pd.DataFrame()
required_revenue_columns = {
    "quantity", "unit_price", "discount_pct", "net_revenue"
}

if required_revenue_columns.issubset(sales.columns):
    revenue_check = sales.copy()
    for col in required_revenue_columns:
        revenue_check[col] = pd.to_numeric(revenue_check[col], errors="coerce")
    revenue_check["hypothetical_revenue"] = (
        revenue_check["quantity"] * revenue_check["unit_price"] *
        (1 - revenue_check["discount_pct"] / 100)
    )
    revenue_check["revenue_difference"] = (
        revenue_check["net_revenue"] - revenue_check["hypothetical_revenue"]
    )
    revenue_check["matches_within_1_cent"] = np.isclose(
        revenue_check["net_revenue"], revenue_check["hypothetical_revenue"],
        atol=0.01, equal_nan=False
    )
    print("Rows tested:", len(revenue_check))
    print("Rows matching hypothetical formula:",
          int(revenue_check["matches_within_1_cent"].sum()))
    print("Rows not matching hypothetical formula:",
          int((~revenue_check["matches_within_1_cent"]).sum()))
    display(revenue_check[[
        "quantity", "unit_price", "discount_pct", "net_revenue",
        "hypothetical_revenue", "revenue_difference",
        "matches_within_1_cent"
    ]].head(20))
else:
    print("Expected revenue columns not all present; inspect sales.columns.")
```

```python
#Consolidated data-quality summary
quality_summary = pd.DataFrame([
    {
        "Dataset": name, "Rows": len(df), "Columns": len(df.columns),
        "Missing Values": int(df.isna().sum().sum()),
        "Duplicate Entire Rows": int(df.duplicated().sum())
    }
    for name, df in {
        "Customer master": customer,
        "Sales transactions": sales,
        "Behaviour events": events
    }.items()
])
display(quality_summary)
```

```python
#Export evidence files
output_dir = Path.cwd() / "Activity_1_Outputs"
output_dir.mkdir(exist_ok=True)

exports = {
    "Activity_1_Data_Quality_Summary.csv": quality_summary,
    "Activity_1_Candidate_Key_Assessment.csv": candidate_keys,
    "Activity_1_Relationship_Summary.csv": relationship_summary,
    "Activity_1_Unmatched_Customer_IDs.csv": unmatched_ids,
    "Activity_1_Date_Range_Summary.csv": date_ranges,
    "Activity_1_Categorical_Profile.csv": categorical_summary
}

for filename, frame in exports.items():
    frame.to_csv(output_dir / filename, index=False)

for name, df in {
    "Customer_Master": customer,
    "Sales_Transactions": sales,
    "Behaviour_Events": events
}.items():
    missing_value_report(df).to_csv(
        output_dir / f"{name}_Missing_Values.csv"
    )

if not date_issues.empty:
    date_issues.to_csv(output_dir / "Activity_1_Date_Issues.csv", index=False)
if not tenure_mismatch.empty:
    tenure_mismatch.to_csv(
        output_dir / "Activity_1_Tenure_Mismatches.csv", index=False
    )
if not revenue_check.empty:
    revenue_check.to_csv(
        output_dir / "Activity_1_Revenue_Check.csv", index=False
    )

print("Evidence files saved to:", output_dir.resolve())
for path in sorted(output_dir.glob("*.csv")):
    print("-", path.name)
```
